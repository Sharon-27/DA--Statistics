```python
A) Calculation of First Quartile (Q1) and Third Quartile (Q3)
First Quartile (Q1):

Using the formula:

Position = \((n-1)\times p\)
Q1\ Position=(8-1)\times0.25=1.75
The position lies between the 1st and 2nd values (6 and 7).
Q1=6+0.75(7-6)=6.75

Therefore, Q1 = 6.75

Third Quartile (Q3):

Q3\ Position=(8-1)\times0.75=5.25


This position falls between the 5th and 6th values (10 and 12).

Q3=10+0.25(12-10)=10.5

Therefore, Q3 = 10.5


(B) Calculation of Interquartile Range (IQR)
The IQR represents the difference between the third and first quartiles.

IQR=Q3-Q1

IQR=10.5-6.75=3.75
Therefore, IQR = 3.75


(C) Calculation of Lower and Upper Fences
The fences help us identify values that may be outliers.

Lower Fence:
Q1-1.5(IQR)

=6.75-1.5(3.75)=1.125
Upper Fence:

Q3+1.5(IQR)
=10.5+1.5(3.75)=16.125
Therefore:
- Lower Fence = 1.125
- Upper Fence = 16.125


(D) Identification of Outliers
An observation is considered an outlier if it is smaller than the lower fence or greater than the upper fence.

Given dataset:

4, 6, 7, 8, 9, 10, 12, 30

After comparing all the values with the calculated fences, we can see that 30 exceeds the upper fence of 16.125.

Therefore, 30 is the only outlier in the given dataset.
```
